```python
x = datas.drop(columns="Target")
y = datas["Target"]
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.25,random_state=10)
```

```python
print(x.shape,x_train.shape,x_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(x_train,y_train)
```

```python
from sklearn.metrics import r2_score,accuracy_score,classification_report
```

```python
accuarcy_test = model.predict(x_test)
accuracy_scores = accuracy_score(accuarcy_test,y_test)

print("The Accuracy_score_test",accuracy_scores)
```

```python
R2_score = model.predict(x_test)
R2_scores = r2_score(R2_score,y_test)

print("The R2_Score are",R2_scores)
```

```python
classification_report_scores = model.predict(x_test)

classificationn = classification_report(classification_report_scores,y_test)

print("The Classificationnn Score Are " , classificationn)
```
